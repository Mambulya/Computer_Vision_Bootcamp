# __Каналы цифрового изображения__
# __Гистограмма изображения__

```python
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np
import io
```

```python
src_pic = "../pics/cat_and_grass.webp"
```

## 1) Замена каналов. Методы Getpixel, putpixel

```python
img = Image.open(src_pic)
```

Замена каналов:

```python
for x in range(img.size[0]):
    for y in range(img.size[1]):
        R, G, B = img.getpixel((x, y))
        img.putpixel((x, y), (B, R, G))

img.show()
```

## 2) Сборка изображения из известных каналов изображения

Разбиение на каналы:

```python
img1 = Image.open("../pics/laying_cat.jpeg")
R, G, B = img1.split()
```

Сборка каналов в изображение:

```python

res = Image.merge(mode="RGB", bands = (R, G, B))
res.show()
```

## 3) Построение диаграммы

```python
f = open(src_pic, "rb").read()
img = Image.open(io.BytesIO(f))
img.convert("L")
```

```python

data = np.array(img)
plt.figure()
plt.hist(data.flatten(), 128)
plt.show()
```
