# МАНИПУЛИРОВАНИЕ ИЗОБРАЖЕНИЕМ

```python
from PIL import Image
```

```python
src_pic = "../pics/strawberries.webp"
```

```python
img = Image.open(src_pic)
img_copy = img.copy()
```

```python
img_copy.show()
```

```python
img_copy.size
```

## Измените размер изображения пропорционально

```python
img_copy.thumbnail((1280, 900))
img_copy.size
```

Стоит заметить, что ширина стала не 900, а 800, так как метод изменяет размер _пропорционально_

## Создание версии изображения указанного размера

Необязательно пропорционально измкеняется размер:

```python
new_size = (1280, 900)
new_img = img.resize(new_size)
new_img.size
```

## Cropping

### Вырезка области изображения и изменение его размеров

Метод crop - [X_left_upper, Y_left_upper, X_right_down, Y_right_down]
<pre>
*-------------------
|                   |
|                   |
|                   |
|                   |
--------------------*
</pre>

```python
cropped_size = [img.size[0]// 4, img.size[1] // 2, img.size[0] * 2.5 // 4, img.size[1] * 9 // 10]
```

```python
img_cropped = img.crop(cropped_size)
img_cropped.load()      # преобразуем в изображение
img_cropped.show()
```

При этом размеры исходного изображения не меняютются:

```python
img.size
```

## Изменение размера изображения

Получим тот же результат без load():

```python
new_cropped_img = img_cropped.resize((int(cropped_size[2] - cropped_size[0]), int(cropped_size[3] - cropped_size[1])))
#                                                    x                                  y
new_cropped_img.show()
```

```python
new_cropped_img.size
```

## Вращение изображения

```python
img_rotated = img.rotate(90)
img_rotated.show()
```

```python
print(img.size)         # размер изображения до поворота
print(img_rotated.size) # размер изображения после поворота
```

Можно заметить, что хоть картинка повернулась, размеры остались теми же

```python
img_rotated = img.rotate(45, expand=True)   # уголки не обрезаются
img_rotated.show()
```

```python
print(img_rotated.size)
```

Изображение вошло полностью (без вырезанных уголков) => размер увеличился

## Получите горизонтальный и зеркальный образ изображения

Зеркальный образ отностительно вертикальной линии:

```python
img_transposed = img.transpose(Image.FLIP_LEFT_RIGHT)   # левая и правая сторона меняются местами
img_transposed.show()
```

Зеркальный образо относительно горизонтальной линии:

```python
img_transposed = img.transpose(Image.FLIP_TOP_BOTTOM)
img_transposed.show()
```

```python
img_transposed = img.transpose(Image.ROTATE_180)
img.show()
```
