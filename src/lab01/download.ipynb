Ознакомление с основными методами загрузки 
изображений и получение общей информации о нем с использованием внешней 
библиотеки PIL.

```python
import pandas as pd
from PIL import Image
import io
```

```python
pic_path = "../pics/laying_cat.jpeg"
```

### Задание 1: Загрузка изображения

```python
img = Image.open(pic_path)
img.show()
```

Конвертация в бинарный режим:

```python
img.convert('L')
img.show()
```

Или же можно сразу открыть в бинарном виде:

```python
file = open(pic_path, "rb")
img = Image.open(file)
img.show()
file.close()
```

Картинка ЧБ:

```python
f = open(pic_path, "rb").read() 
img = Image.open(io.BytesIO(f))
img = img.convert("L")
img.show()
```

### Задание 2: Получение информации об изображении

```python
img = Image.open(pic_path)
```

```python
# размер изображения
print(img.size)
# расширение
print(img.format)
# система расцветки
print(img.mode)
# координаты
print(img.getbbox())
# информация об изображении
print(img.info)
```
