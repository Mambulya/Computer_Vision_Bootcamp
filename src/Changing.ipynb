```python
from PIL import Image
import pylab    # Numpy + Matplotlib
```

```python
src_pic = "../pics/dog_and_flowers.webp"
```

```python
img = Image.open(src_pic)
```

```python
x, y = img.size
x, y
```

## 1) Закраска (заливка) части и всего изображения заданным цветом

Закрасте часть картинкии черным цветом

```python
img1 = img.copy()
img1.paste((0, 0, 0), (x * 2 // 5 - 50, 220, x * 3 // 5 - 50, 370))
img1.show()
```

### Залейте все изображение зеленым цветом

```python
img_green = img.copy()
img_green.paste((0, 255, 0), (0, 0, x, y))
img_green.show()
```

## 2) Комбинированные манипуляции на изображении

### Загрузите изображение, создайте его уменьшенную, а затем вставьте ее в исходное изображение, причем вокруг вставленного изображения отобразите рамку красного цвета.

```python
img2 = img.copy()
small_pic = img.resize((x // 4, y // 4))
img2.paste((128, 0, 0), (200, 200, 200 + small_pic.size[0] + 20, 200 + small_pic.size[1] + 20))
img2.paste(small_pic, (210, 210))
img2.show()
```

### Выведите белую полупрозрачную горизонтальную полосу высотой 100 пикселов

```python
img_white = img.copy()
white_line = Image.new("RGB", (x, 100), (255, 255, 255))
mask = Image.new("L", (x, 100), 65)
img_white.paste(white_line, (0, 0), mask)
img_white.show()
```

### Копируйте и вставьте повернутую часть изображения

```python
img3 = img.copy()
img_rotated = img3.crop((x * 2 // 5 - 50, 220, x * 3 // 5 - 50, 370))
img3.paste(img_rotated.transpose(Image.ROTATE_90), (500, 500))
img3.show()
```
